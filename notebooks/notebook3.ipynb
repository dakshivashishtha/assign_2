```python
import pandas as pd
import numpy as np
```

```python
DATA_PATH = "../data/processed/nykaa_campaign_cleaned.csv"

df = pd.read_csv(DATA_PATH)

print(f"Shape of dataset: {df.shape[0]} rows x {df.shape[1]} columns")
df.head()
```

```python
print("Columns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

```python
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
```

```python
df["Date"].dtype
```

```python
df["Spend"] = df["Acquisition_Cost"] * df["Conversions"]

df[["Acquisition_Cost", "Conversions", "Spend"]].head()
```

```python
df["CTR"] = np.where(
    df["Impressions"] > 0,
    (df["Clicks"] / df["Impressions"]) * 100,
    np.nan
)

df["Conversion_Rate"] = np.where(
    df["Clicks"] > 0,
    (df["Conversions"] / df["Clicks"]) * 100,
    np.nan
)

df["CPC"] = np.where(
    df["Clicks"] > 0,
    df["Spend"] / df["Clicks"],
    np.nan
)

df["CPA"] = np.where(
    df["Conversions"] > 0,
    df["Spend"] / df["Conversions"],
    np.nan
)

df["ROAS"] = np.where(
    df["Spend"] > 0,
    df["Revenue"] / df["Spend"],
    np.nan
)

df["CPL"] = np.where(
    df["Leads"] > 0,
    df["Spend"] / df["Leads"],
    np.nan
)
```

```python
kpi_columns = [
    "Spend",
    "CTR",
    "Conversion_Rate",
    "CPC",
    "CPA",
    "ROAS",
    "CPL"
]

display(df[kpi_columns].describe().T)
```

```python
display(df[kpi_columns].head(10))
```

```python
df["Lead_Rate"] = np.where(
    df["Clicks"] > 0,
    (df["Leads"] / df["Clicks"]) * 100,
    np.nan
)

df["Lead_to_Conversion_Rate"] = np.where(
    df["Leads"] > 0,
    (df["Conversions"] / df["Leads"]) * 100,
    np.nan
)

df["Overall_Funnel_Efficiency"] = np.where(
    df["Impressions"] > 0,
    (df["Conversions"] / df["Impressions"]) * 100,
    np.nan
)
```

```python
funnel_features = [
    "Lead_Rate",
    "Lead_to_Conversion_Rate",
    "Overall_Funnel_Efficiency"
]

display(df[funnel_features].describe().T)
```

```python
# Campaign Intensity Features
df["Impressions_per_Day"] = np.where(
    df["Duration"] > 0,
    df["Impressions"] / df["Duration"],
    np.nan
)

df["Clicks_per_Day"] = np.where(
    df["Duration"] > 0,
    df["Clicks"] / df["Duration"],
    np.nan
)

df["Leads_per_Day"] = np.where(
    df["Duration"] > 0,
    df["Leads"] / df["Duration"],
    np.nan
)

df["Conversions_per_Day"] = np.where(
    df["Duration"] > 0,
    df["Conversions"] / df["Duration"],
    np.nan
)

df["Spend_per_Day"] = np.where(
    df["Duration"] > 0,
    df["Spend"] / df["Duration"],
    np.nan
)
```

```python
intensity_features = [
    "Impressions_per_Day",
    "Clicks_per_Day",
    "Leads_per_Day",
    "Conversions_per_Day",
    "Spend_per_Day"
]

display(df[intensity_features].describe().T)
```

```python
# Date-based features
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Quarter"] = df["Date"].dt.quarter
df["Day_of_Week"] = df["Date"].dt.dayofweek
df["Is_Weekend"] = (df["Day_of_Week"] >= 5).astype(int)
```

```python
display(
    df[
        [
            "Date",
            "Year",
            "Month",
            "Quarter",
            "Day_of_Week",
            "Is_Weekend"
        ]
    ].head(10)
)
```

```python
df["Channel_Used"].value_counts().head(20)
```

```python
df["Channel_Count"] = (
    df["Channel_Used"]
    .fillna("")
    .str.split(",")
    .apply(lambda channels: len([c.strip() for c in channels if c.strip()]))
)

df["Is_Multichannel"] = (df["Channel_Count"] > 1).astype(int)
```

```python
channel_dummies = (
    df["Channel_Used"]
    .fillna("")
    .str.get_dummies(sep=",")
)

channel_dummies.columns = [
    "Channel_" + col.strip().replace(" ", "_")
    for col in channel_dummies.columns
]

df = pd.concat([df, channel_dummies], axis=1)

print("Separate channel features created.")
print(channel_dummies.columns.tolist())
```

```python
df
```

```python
categorical_columns = [
    "Campaign_Type",
    "Target_Audience",
    "Language",
    "Customer_Segment"
]

df = pd.get_dummies(
    df,
    columns=categorical_columns,
    drop_first=True,
    dtype=int
)
```

```python
missing_features = df.isnull().sum()
```

```python
missing_features = missing_features[missing_features > 0].sort_values(
    ascending=False
)
```

```python
display(missing_features)
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

infinite_counts = np.isinf(df[numeric_columns]).sum()

display(
    infinite_counts[infinite_counts > 0]
)
```

```python
# Remove duplicate columns
df = df.loc[:, ~df.columns.duplicated()]

print("Shape after removing duplicate columns:", df.shape)
```

```python
duplicate_columns = df.columns[df.columns.duplicated()]

print("Duplicate columns:")
print(duplicate_columns.tolist())
```

```python
channel_columns = [
    col for col in df.columns
    if col.startswith("Channel_")
]

print(channel_columns)
```

```python
print("Final shape:", df.shape)

print("\nFinal columns:")
print(df.columns.tolist())
```

```python
# removing raw data as we odnt need it anymore
df_features = df.drop(
    columns=["Date", "Channel_Used"]
)

print("Feature-engineered dataset shape:", df_features.shape)
```
